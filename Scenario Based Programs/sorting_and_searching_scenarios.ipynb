**Scenario 1: E-Commerce Product Catalog & Dynamic Filtering**

```python
import random, time


def insertion_sort(collection):
    for i in range(1, len(collection)):
        element = collection[i]
        for j in range(i - 1, -1, -1):
            if element < collection[j]:
                collection[j + 1] = collection[j]
            else:
                collection[j + 1] = element
                break
        else:
            collection[0] = element
    return collection


prices = [round(random.uniform(50, 5000), 2) for i in range(5000)]

start_time = time.time()
sorted_price = insertion_sort(prices)
end_time = time.time()

print(
    f"Sorted 5,000 prices using Insertion Sort in {end_time - start_time:.4f} seconds."
)
print("Top 5 lowest prices:", sorted_price[:5])
```

**Scenario 2: GPS Navigation & Nearest Location Lookup**

```python
route_milestones = list(range(0, 1000000, 2))
target_location = 499998


def binary_search(required_item, collection, low, high, step=1):
    if low > high:
        return "Not found", step

    else:
        mid = (low + high) // 2

        if collection[mid] == required_item:
            return mid, step

        elif collection[mid] < required_item:
            return binary_search(required_item, collection, mid + 1, high, step + 1)

        elif collection[mid] > required_item:
            return binary_search(required_item, collection, low, mid - 1, step + 1)


index, steps = binary_search(
    target_location, route_milestones, 0, len(route_milestones) - 1
)
print(f"Binary Search found {target_location} at index {index} in just {steps} steps!")
```

**Scenario 3: Memory-Constrained Embedded IoT Devices (Smart Grid Meters)**

```python
sensor_readings = [230.2, 228.5, 231.0, 229.8, 230.5, 227.1]

print("Raw sensor batch:", sensor_readings)


def selection_sort(collection):
    for i in range(1, len(collection)):
        min_index = i - 1
        for j in range(i, len(collection)):
            if collection[j] < collection[min_index]:
                min_index = j

        collection[i - 1], collection[min_index] = (
            collection[min_index],
            collection[i - 1],
        )
    return collection


sorted_readings = selection_sort(sensor_readings)
print("Sorted telemetry (In-place):", sorted_readings)
```

**Scenario 4: Real-Time Log File Processing & Anomaly Detection**

```python
firewall_logs = [1021, 1025, 1030, 1028, 1035, 1040]
print(f"Before Sorting: {firewall_logs}")


def insertion_sort(collection):
    for i in range(1, len(collection)):
        element = collection[i]
        j = i - 1

        while j >= 0 and element < collection[j]:
            collection[j + 1] = collection[j]
            j -= 1
        collection[j + 1] = element
    return collection


print(f"After insertion sort: {insertion_sort(firewall_logs)}")
```
