**Binary Search using Recursion**

```python
def binary_search(required_item, collection, low, high):
    if low > high:
        return f"{required_item} does not exist in the collection"

    else:
        mid = (low + high) // 2

        if collection[mid] == required_item:
            return f"Found {required_item} at index {mid}"

        elif collection[mid] < required_item:
            return binary_search(required_item, collection, mid + 1, high)

        elif collection[mid] > required_item:
            return binary_search(required_item, collection, low, mid - 1)


collection = eval(input("Enter collection [Should be already sorted]: "))
required_item = eval(input("Enter item to be searched: "))
low = 0
high = len(collection) - 1
print(binary_search(required_item, collection, low, high))
```

**Binary Search using While loop**

```python
def binary_search(required_item, collection):
    low, high = 0, len(collection) - 1

    while low <= high:
        mid = (low + high) // 2

        if collection[mid] == required_item:
            return f"Found {required_item} at index {mid}"

        elif collection[mid] < required_item:
            low = mid + 1

        elif collection[mid] > required_item:
            high = mid - 1

    else:
        return f"{required_item} does not exist in the collection"


collection = eval(input("Enter collection [Should be already sorted]: "))
required_item = eval(input("Enter item to be searched:"))
print(binary_search(required_item, collection))
```
