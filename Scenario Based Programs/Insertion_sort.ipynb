**Insertion Sort - For loop**

```python
def insertion_sort(collection):
    for i in range(1, len(collection)):
        element = collection[i]
        for j in range(i - 1, -1, -1):
            if element < collection[j]:
                collection[j + 1] = collection[j]
            else:
                collection[j + 1] = element
                break
        else:
            collection[0] = element
    return collection


collection = eval(input("Enter collection: "))
print(insertion_sort(collection))
```

**Insertion Sort - While loop**

```python
def insertion_sort(collection):
    for i in range(1, len(collection)):
        element = collection[i]
        j = i - 1
        while j >= 0 and element < collection[j]:
            collection[j + 1] = collection[j]
            j -= 1
        collection[j + 1] = element
    return collection


collection = eval(input("Enter collection: "))
print(insertion_sort(collection))
```
